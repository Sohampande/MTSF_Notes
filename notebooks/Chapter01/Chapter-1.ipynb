```python
import matplotlib.pyplot as plt

def plot_time_series(time, values, title):
    plt.figure(figsize=(10, 6))
    plt.plot(time, values, color="blue")
    plt.title(title)
    plt.xlabel("Time")
    plt.ylabel("Value")
    plt.grid(True)
    plt.show()
```

```python
import numpy as np
time = np.arange(200)
values = np.random.randn(200) * 100
plot_time_series(time, values, "White noise")
```
